### 4. Completa la clasificación
Se asume una escala de notas de 0.0 a 5.0, con los siguientes rangos: Excelente ≥ 4.5, Muy buen desempeño ≥ 3.5, Aprobado ≥ 3.0, y No aprobado por debajo de 3.0.

```python
nota = 4.2
if nota >= 4.5:
    print("Excelente")
elif nota >= 3.5:
    print("Muy buen desempeño")
elif nota >= 3.0:
    print("Aprobado")
else:
    print("No aprobado")
```

### 5. Corrige el orden
El problema original evaluaba primero `nota >= 3.0`, que siempre es verdadera para notas altas, por lo que nunca se alcanzaba el caso "Excelente". La condición más específica (más alta) debe ir primero.

```python
nota = 4.8
if nota >= 4.5:
    print("Excelente")
elif nota >= 3.0:
    print("Aprobado")
```
