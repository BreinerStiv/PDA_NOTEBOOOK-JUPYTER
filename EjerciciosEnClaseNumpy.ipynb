```python
import pandas as pd

df_ventas = pd.DataFrame(datos_ventas)

df_ventas["fecha"] = pd.to_datetime(df_ventas["fecha"])


numero_filas = df_ventas.shape[0]          # 16​

numero_columnas = df_ventas.shape[1]       # 9​

columnas_disponibles = list(df_ventas.columns)

df_ventas.head()
```

```python
columnas_analisis = ["fecha", "producto", "cantidad", "sede"]

mascara = (

    (df_ventas["sede"] == "Centro")

    & (df_ventas["categoria"] == "Almuerzo")

)

ventas_centro_almuerzo = df_ventas.loc[mascara, columnas_analisis].copy()


unidades_centro_almuerzo = ventas_centro_almuerzo["cantidad"].sum()
```

```python
df_ventas["ingreso"] = df_ventas["cantidad"] * df_ventas["precio_unitario"]

df_ventas["costo_total"] = df_ventas["cantidad"] * df_ventas["costo_unitario"]

df_ventas["margen_bruto"] = df_ventas["ingreso"] - df_ventas["costo_total"]

df_ventas["porcentaje_margen"] = (

    df_ventas["margen_bruto"] / df_ventas["ingreso"] * 100

)
```

```python
kpi_producto = (

    df_ventas

    .groupby("producto")
    .agg(

        unidades=("cantidad", "sum"),
        ingreso=("ingreso", "sum"),

        margen=("margen_bruto", "sum"),

    )

    .reset_index()

)


kpi_producto["porcentaje_margen"] = (

    kpi_producto["margen"] / kpi_producto["ingreso"] * 100

).round(2)
```

```python
kpi_producto = kpi_producto.sort_values("ingreso", ascending=False)



producto_lider = kpi_producto.iloc[0]["producto"]   # Bowl Andino​

# alternativa sin ordenar​
indice = kpi_producto["ingreso"].idxmax()

producto_lider = kpi_producto.loc[indice, "producto"]


```

```python
def resumen_ejecutivo(df):
    ingreso = df["ingreso"].sum()

    margen = df["margen_bruto"].sum()


    s1 = set(df.loc[df["semana"] == 1, "cliente_id"])

    s2 = set(df.loc[df["semana"] == 2, "cliente_id"])

    recurrentes = s1 & s2



    return {

        "total_ingresos": ingreso,

        "margen_bruto": margen,

        "porcentaje_margen": round(margen / ingreso * 100, 2),

        "ticket_promedio": round(ingreso / len(df), 2),

        "clientes_recurrentes": len(recurrentes),

        "tasa_retencion": round(len(recurrentes) / len(s1) * 100, 2),

    }
```
