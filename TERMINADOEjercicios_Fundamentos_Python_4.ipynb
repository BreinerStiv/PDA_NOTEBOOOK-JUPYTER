### 9. Un método que modifica

```python
class Estudiante:
    def __init__(self, nombre, nota):
        self.nombre = nombre
        self.nota = nota

    def aprobo(self):
        return self.nota >= 3.0

    def subir(self, puntos):
        self.nota = self.nota + puntos
        return self.nota

e = Estudiante("Luis", 3.0)
print(e.subir(0.5))
```

### 10. Método con parámetro

```python
class Estudiante:
    def __init__(self, nombre, nota):
        self.nombre = nombre
        self.nota = nota

    def aprobo(self):
        return self.nota >= 3.0

    def subir(self, puntos):
        self.nota = self.nota + puntos
        return self.nota

    def promedio_con(self, otra):
        return (self.nota + otra) / 2

e = Estudiante("Sara", 3.0)
print(e.promedio_con(4.0))
```

### 14. Validar antes de modificar

```python
class Cuenta:
    def __init__(self, saldo):
        self.__saldo = saldo

    def consultar(self):
        return self.__saldo

    def consignar(self, valor):
        if valor <= 0:
            return "Valor inválido"
        self.__saldo = self.__saldo + valor
        return self.__saldo

c = Cuenta(100)
print(c.consignar(-20))
print(c.consignar(50))
```

### 15. `__str__`: una vista legible

```python
class Estudiante:
    def __init__(self, nombre, nota):
        self.nombre = nombre
        self.nota = nota

    def __str__(self):
        return f"{self.nombre}: {self.nota}"

print(Estudiante("Ana", 4.2))
```

### 20. La clase

```python
class Estudiante:
    def __init__(self, nombre, nota):
        self.nombre = nombre
        self.nota = nota

    def __str__(self):
        return f"{self.nombre}: {self.nota}"
class Curso:
    def __init__(self, nombre):
        self.nombre = nombre
        self.estudiantes = []

    def inscribir(self, estudiante):
        self.estudiantes.append(estudiante)

    def promedio(self):
        suma = 0
        for e in self.estudiantes:
            suma = suma + e.nota
        return round(suma / len(self.estudiantes), 2)

c = Curso("Analítica")
c.inscribir(Estudiante("Ana", 4.2))
c.inscribir(Estudiante("Luis", 3.1))
print(c.promedio())
```
